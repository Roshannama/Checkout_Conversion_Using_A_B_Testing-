```python
import pandas as pd
import numpy as np

users = pd.read_csv("../data/raw/users.csv")
products = pd.read_csv("../data/raw/products.csv")
sessions = pd.read_csv("../data/raw/sessions.csv")
events = pd.read_csv("../data/raw/events.csv")
orders = pd.read_csv("../data/raw/orders.csv")
experiment = pd.read_csv("../data/raw/experiment_assignment.csv")
```

```python
print(users.shape)
print(products.shape)
print(sessions.shape)
print(events.shape)
print(orders.shape)
print(experiment.shape)
```

```python
for name, df in {
    "users": users,
    "products": products,
    "sessions": sessions,
    "events": events,
    "orders": orders,
    "experiment": experiment
}.items():
    print("=" * 50)
    print(name)
    print("=" * 50)
    df.info()
    display(df.describe(include="all"))
    display(df.head())
```

```python
print("Users:", users["user_id"].nunique(), users.shape[0])
print("Products:", products["product_id"].nunique(), products.shape[0])
print("Sessions:", sessions["session_id"].nunique(), sessions.shape[0])
print("Events:", events["event_id"].nunique(), events.shape[0])
print("Orders:", orders["order_id"].nunique(), orders.shape[0])
print("Assignments:", experiment["assignment_id"].nunique(), experiment.shape[0])
```

```python
invalid_sessions = sessions[
    ~sessions["user_id"].isin(users["user_id"])
]

print(len(invalid_sessions))
```

```python
invalid_events = events[
    ~events["session_id"].isin(sessions["session_id"])
]

print(len(invalid_events))
```

```python
invalid_orders_users = orders[
    ~orders["user_id"].isin(users["user_id"])
]

print(len(invalid_orders_users))
```

```python
invalid_orders_sessions = orders[
    ~orders["session_id"].isin(sessions["session_id"])
]

print(len(invalid_orders_sessions))
```

```python
invalid_orders_products = orders[
    ~orders["product_id"].isin(products["product_id"])
]

print(len(invalid_orders_products))
```

```python
users["signup_date"] = pd.to_datetime(
    users["signup_date"],
    errors="coerce"
)

sessions["session_start"] = pd.to_datetime(
    sessions["session_start"],
    errors="coerce"
)

sessions["session_end"] = pd.to_datetime(
    sessions["session_end"],
    errors="coerce"
)

events["event_time"] = pd.to_datetime(
    events["event_time"],
    errors="coerce"
)

orders["purchase_time"] = pd.to_datetime(
    orders["purchase_time"],
    errors="coerce"
)

products["launch_date"] = pd.to_datetime(
    products["launch_date"],
    errors="coerce"
)
```
